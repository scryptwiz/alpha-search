```python
import json
from alpha_search.search import search_alpha


question = "What is Nvidia?" 
```

```python
from dotenv import load_dotenv
from openai import OpenAI

from alpha_search.llm import extract_search_query

load_dotenv()
client = OpenAI()
model = "gpt-4o-mini"

query = extract_search_query(question, client=client, model=model)
print(f"Extracted query: {query}")
```

```python
# Search Alpha for articles related to the query
alpha_result = search_alpha(query)
print(f"Found {len(alpha_result)} results for query: {query}")
print(json.dumps(alpha_result, indent=4))
```

```python
from alpha_search.fetch import fetch_top_results

# Fetch the top results from the Alpha search
pages = fetch_top_results(alpha_result, n=3)

for page in pages:
    print(page["score"], page["title"])
    print(page["url"])
    print(page["text"][:400])
    print(len(page["text"]), "chars")
```

```python
from alpha_search.llm import answer_from_sources

answer = answer_from_sources(
    question,
    pages,
    client=client,
    model=model,
)
print(answer)
```
