```python
import json
from alpha_search.search import search_alpha


query = "NVIDIA"
question = "Why did Nvidia acqui-hire Groq?" 

result = search_alpha(query)
print(f"Found {len(result)} results for query: {query}")
print(json.dumps(result, indent=4))
```

```python
from alpha_search.fetch import fetch_top_results


pages = fetch_top_results(result, n=3)

for page in pages:
    print(page["score"], page["title"])
    print(page["url"])
    print(page["text"][:400])
    print(len(page["text"]), "chars")
```

```python
context_blocks = []
for i, page in enumerate(pages, start=1):
    context_blocks.append(
        f"[{i}] {page['title']}\n"
        f"URL: {page['url']}\n\n"
        f"{page['text'][:6000]}"  # cap length to save tokens/$
    )
context = "\n\n---\n\n".join(context_blocks)
print(context)
```

```python
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv()
client = OpenAI()

model = "gpt-4o-mini"

messages = [
    {
        "role": "system",
        "content": (
            "You are a helpful retrieval assistant for Alpha Search. \n"
            "Your job is to answer the user's question using ONLY the source articles provided in the message. Do not use outside knowledge. Do not invent facts, quotes, or links. \n"
            "If the sources do not contain enough information, say you don't know and suggest a better search query. \n"
            "When you use information from a source, ground the answer in that source. At the end of every reply, add a Sources section listing each article you used in this format:"
            "Sources: \n"
            "- Title — URL \n"
            "Only list sources you actually used. Prefer concise, direct answers."
        ),
    },
    {
        "role": "user",
        "content": (
            f"Question: {question}\n\n"
            f"Source articles:\n\n"
            f"{context}"
        ),
    }
]
response = client.chat.completions.create(
    model=model,
    messages=messages,
)
print(response.choices[0].message.content)
```
