```python
import json

question = "Why did Nvidia acqui-hire Groq?"
```

```python
from dotenv import load_dotenv
from openai import OpenAI

from alpha_search.llm import extract_search_query

load_dotenv()
client = OpenAI()
model = "gpt-4o-mini"

query = extract_search_query(question, client=client, model=model)
print(f"Extracted query: {query}")
```

```python
from alpha_search.search import search_alpha
from alpha_search.search_bm25 import search_bm25

bm25_results = search_bm25(query)

print("Basic Search", search_alpha(query)[:3])

print("\nv2")
print(json.dumps(bm25_results[:5], indent=2, ensure_ascii=False))
```

```python
from alpha_search.fetch import fetch_top_results

bm25_pages = fetch_top_results(results=bm25_results, n=3)
for page in bm25_pages:
    status = "ok" if page.get("ok") else f"FAIL: {page.get('error')}"
    print(page["score"], status, page["title"])
    print(page["url"])
    print((page.get("text") or "")[:300])
    print("---")
```

```python
from alpha_search.llm import answer_from_sources


answer = answer_from_sources(
    question,
    bm25_pages,
    client=client,
    model=model,
)
print(answer)
```
